```python
import matplotlib.pyplot as plt
x =[1,2,3]
y=[2,4,8]
plt.plot(x,y)
```

```python
import matplotlib
matplotlib.rcParams["font.family"] = 'HYGungSo-Bold'
matplotlib.rcParams["font.size"] = 15
matplotlib.rcParams["axes.unicode_minus"]=False

plt.plot(x,y)
plt.title("꺾은선 그래프")
```

```python
plt.plot(x,y)
plt.title("꺾은선 그래프", fontdict ={"family":"HYGungSo-Bold","size": 20}) #폰트 개별 설정
```

```python
plt.plot(x,y)
plt.xlabel("X축", color ="red",loc = "right") #x축에 라벨과 색깔 붙히기 & 위치 조정하기 
# left center right
plt.ylabel("Y축", color ="#00aa00",loc = "top")
# top center bottom
```

```python
plt.plot(x,y)
plt.xticks([1,2,3])
plt.yticks([3,6,9,12])
plt.show()
```
