# 04. 스타일

```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams["font.family"] = "Malgun Gothic"
matplotlib.rcParams["font.size"] = 15
matplotlib.rcParams["axes.unicode_minus"] =False
```

```python
x = [10,20,40]
y = [2,4,10]
```

```python
plt.plot(x,y,linewidth=10) # 라인 두께 설정
```

```python
plt.plot(x,y , marker ="o") # 마커를 o로 표
```

## Marker

```python
plt.plot(x,y , marker ="o") # 마커를 o로 표
```

```python
plt.plot(x,y, marker = "o", linestyle ="None") ##라인 없애기 
```

```python
plt.plot(x,y, marker ="v", markersize=10) ## 마커를 v로
```

```python
plt.plot(x,y,marker = "X", markersize = 10) 
##matplotlib.markers 를 구글링하면 필요한 마커들 볼 수 있음.
```

```python
plt.plot(x,y, marker ="X", markersize = 10, markeredgecolor = "red") 
# 뭔지 알겠지??
```

```python
plt.plot(x,y,marker ="o", markersize = 20, markerfacecolor = "yellow")
```

## 선 스타일

```python
plt.plot(x,y,linestyle = "dotted") # solid , dotted, dashed, dashdot
```

```python
plt.plot(x,y,linestyle = "dashdot")
```

## 색깔

```python
plt.plot(x,y, color ="pink")
```

```python
plt.plot(x,y, color ="#ff0000") # RGB값 
```

## 포맷

```python
plt.plot(x,y,"ro--") # color : r / marker = o / linestyle = --
```

```python
plt.plot(x,y, "bv:")
```

```python
plt.plot(x,y, color = "#ff0000", marker = "o", markersize = 20 , markeredgecolor = "red")
```

## 축약어

```python
plt.plot(x,y, marker ="o", mfc = "red", ms = 10 , mec = "blue", ls = ":")
## matplotlib markerfacecolor mfc 로 쳐보면 축약어가 나온다.
```

## 투명도

```python
plt.plot(x,y, marker ="o", mfc = "red", ms = 10 , alpha = 0.3 ) # alpha : 투명도 0~1
```

## 그래프 크기

```python
plt.figure(figsize =(10,5)) ## 그래프의 크기를 키우는
plt.plot(x,y)
```

```python
plt.figure(figsize =(1,1)) ## 그래프의 크기를 결정
plt.plot(x,y)
```

```python
plt.figure(figsize =(10,5), dpi = 200) ## dot per inches 
plt.plot(x,y)
```

## 배경색

```python
plt.figure(facecolor = "yellow")
plt.plot(x,y)
```
