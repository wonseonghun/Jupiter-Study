# 08.막대그래프(기본)

```python
import matplotlib.pyplot as plt
import matplotlib

matplotlib.rcParams["font.family"] = "Malgun Gothic"
matplotlib.rcParams["font.size"] = 15
matplotlib.rcParams["axes.unicode_minus"] =False
```

```python
labels = ["강백호","서태웅","정대만"]
values=[190,187,184]

plt.bar(labels,values)
```

```python
labels = ["강백호","서태웅","정대만"]
values=[190,187,184]

plt.bar(labels,values, color = "pink")
```

```python
labels = ["강백호","서태웅","정대만"]
values=[190,187,184]
colors=["r","b","g"]
plt.bar(labels,values, color = colors, alpha = 0.5)
```

```python
labels = ["강백호","서태웅","정대만"]
values=[190,187,184]

plt.bar(labels,values)
plt.ylim(175, 195) # make y limit
```

```python
plt.bar(labels,values, width = 0.5) # make bars more thin
```

```python
plt.bar(labels, values, width = 0.3)
plt.xticks(rotation = 45)
plt.yticks(rotation = 45) # it is useful when the words overlap each other
```

```python
labels =["강백호","서태웅","정대만"]
values= [190,187,184]
ticks = ["1번학생","2번학생",'3번학생']

plt.bar(labels, values)
plt.xticks(labels,ticks, rotation= 90) #  change labels as ticks that i made + rotate
```
