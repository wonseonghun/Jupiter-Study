```python
df.shape # row,column 알기 => 데이터가 많을 때 사용
```

```python
N = df.shape[0]
N
```

```python
index = np.arange(N)
index
```

```python
w = 0.25
plt.bar(index-w, df["국어"], width = 0.25, label ="국어")
plt.bar(index, df["영어"], width = 0.25, label ="영어")
plt.bar(index + w, df["수학"], width = 0.25, label = "수학")
plt.legend(ncol=3)  # 가로로
```

```python
plt.figure(figsize = (10,5))
plt.title("학생별 성적")

w = 0.25
plt.bar(index-w, df["국어"], width = 0.25, label ="국어")
plt.bar(index, df["영어"], width = 0.25, label ="영어")
plt.bar(index + w, df["수학"], width = 0.25, label = "수학")
plt.legend(ncol=3)  # 가로로
plt.xticks(index, df["이름"], rotation = 30) # 숫자 대신 이름으로 변경

plt.show()
```
