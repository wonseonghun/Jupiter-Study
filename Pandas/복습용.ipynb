# 9.데이터선택(조건)

조건에 해당하는 데이터 선택

```python
df["키"]>=185
```

```python
filt=(df["키"]>180)
df[filt]
```

```python
df.loc[df["키"]>160,"수학"]
```

```python
df.loc[(df["키"]>=185) & (df["학교"]=="북산고")]
```

```python
df.loc[(df["키"]<200)|(df["학교"]=="능남고")]
```

```python
filt = df["이름"].str.startswith("송")
df[filt]
```

```python
langs=["python","java"]
filt= df["SW특기"].str.lower().isin(langs)
df[filt]
```
